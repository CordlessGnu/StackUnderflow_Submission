# Import modules

```python
import torch
from ultralytics import YOLO
print(torch.cuda.is_available())
```

# Load pretrained model

```python
model = YOLO("Yolo26s.pt")
```

# Train model

```python
model.train(data = "/home/ji/Code/Yolo26 ClashRoyale/Hog2.6 ally - enemy.v3i.yolo26/data.yaml", imgsz = 640, batch = 8, epochs = 200, workers = 4 , device = "0")
```

# Test model on a testing image

```python
model.predict(source="/home/ji/Code/Yolo26 ClashRoyale/Hog2.6 ally - enemy.v1i.yolo26/test/images/frame_0090_045_jpg.rf.e0f9e96dcd18621acaeff478f6bdd6dc.jpg", show = True, save=True)
```
